## 2. The question: decision, action, cost of a wrong call

Markdown
* **Research Question:** When compressing an open-source 7B language model to 4-bit precision, how much reasoning accuracy is sacrificed relative to the massive reduction in memory footprint?
* **Unit of Analysis:** A single open-source 7B parameter model evaluated across standard benchmark test prompts.
* **The Decision:** Whether an engineering team or researcher should deploy a resource-efficient quantized model (4-bit) or stick to a high-resource baseline model (16-bit).
* **The Action:** Deploy the 4-bit quantized model to lower hardware/VRAM infrastructure costs by over 60%, accepting a minor, measurable accuracy trade-off, or maintain the 16-bit baseline where zero tolerance for accuracy loss is required.
* **Cost of a Wrong Call:** Deploying an improperly evaluated quantized model could lead to severe reasoning degradation or hallucinations on complex tasks, causing downstream logic errors. Conversely, refusing to quantize unnecessarily increases server compute costs by more than 2x.
