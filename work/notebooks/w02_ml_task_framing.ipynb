## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

Task Type: Scoring / Regression & Performance Evaluation.

Description: Rather than predicting a simple pass/fail flag (classification), this task evaluates and scores model efficiency trade-offs by predicting exact continuous values—specifically memory consumption (VRAM footprint in MB/GB) and accuracy retention (percentage performance on benchmark evaluation tasks) when applying model quantization.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Target Variables:

1. vram_usage_gb (Continuous float): Peak VRAM memory allocated during model inference.

2. accuracy_retention_score (Continuous float): Evaluation score on test benchmarks relative to the 16-bit unquantized baseline.

Proxy Explanation: Since testing every parameter configuration on live hardware takes time, predicting these target values serves as a proxy for operational feasibility before deploying model weights to memory-constrained edge devices.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Primary Metrics:

VRAM Efficiency: Percentage reduction in peak memory usage (Target: $\ge 60\%$ reduction).

Accuracy Retention: Percentage of original model performance preserved (Target: $\ge 94\%$ retention).

Regression Evaluation Metric: Root Mean Squared Error ($RMSE$) or Mean Absolute Percentage Error ($MAPE$) between predicted resource consumption and real hardware benchmarks.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Why a Fixed Rule Fails:

A fixed rule like "Always quantize to 4-bit to save memory" is too naive because non-linear quantization degradation depends heavily on model architecture, task context length, and group sizes. 4-bit precision might preserve 94.2% accuracy on simple text generation, but crash heavily on complex mathematical reasoning or coding tasks.

Why ML/Data Analysis Beats Fixed Rules:

An ML scoring approach evaluates dynamic trade-off curves across different tasks and precision levels. It allows us to predict the exact threshold where VRAM savings diminish while accuracy degradation accelerates, enabling automated, workload-specific model deployment decisions.
